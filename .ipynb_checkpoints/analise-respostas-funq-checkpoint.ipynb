```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# O CSV exportado pela plataforma usa ';' como separador e UTF-8 com BOM
df = pd.read_csv('Respostas_FunQ_Alunos.csv', sep=';', encoding='utf-8-sig')

# Idade ausente no export aparece como 'N/A'; converte para número (NaN quando ausente)
df['Idade'] = pd.to_numeric(df['Idade'], errors='coerce')

df = df.drop_duplicates()
pd.set_option('display.width', 1000)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)

# Mapeia as 6 dimensões do FunQ (blocos de 3 questões)
dimensoes = {
    'Autonomia': ['q1', 'q2', 'q3'],
    'Desafio': ['q4', 'q5', 'q6'],
    'Prazer': ['q7', 'q8', 'q9'],
    'Imersão': ['q10', 'q11', 'q12'],
    'Barreiras_Sociais': ['q13', 'q14', 'q15'],
    'Estresse': ['q16', 'q17', 'q18'],
}
```

## 📖 Dicionário de Variáveis — Questionário FunQ


### 1. Questões Padrões do FunQ (Q1 a Q18)

| **q2** | **Autonomia** | **A2** | Eu fiz esta atividade porque eu tinha que fazer. | **Sim** |
| **q3** | **Autonomia** | **A3** | Fiz esta atividade porque eu quis. | Não |
| **q4** | **Desafio** | **C1** | Senti que eu era bom(boa) nesta atividade. | Não |
| **q5** | **Desafio** | **C2** | Eu fiz algo novo. | Não |
| **q6** | **Desafio** | **C3** | Fiquei curioso(a). | Não |

| **q8** | **Prazer** | **D2** | Gostaria de fazer algo parecido novamente. | Não |
| **q9** | **Prazer** | **D3** | Senti-me feliz. | Não |
| **q10** | **Imersão** | **I1** | Senti que o tempo passou voando. | Não |
| **q11** | **Imersão** | **I2** | Esqueci da escola. | Não |

> **Nota Metodológica:** As questões marcadas com escore reverso já tiveram seus valores invertidos no pré-processamento do *back-end* ($\text{Novo Valor} = 6 - \text{Valor Antigo}$), garantindo que em todas as colunas uma nota mais alta signifique uma maior intensidade da dimensão avaliada.

```python
# Média de cada dimensão por aluno

# Calcula a média de cada dimensão por resposta/aluno (axis=1 calcula por linha)
for nome_dimensao, colunas in dimensoes.items():
    df[nome_dimensao] = df[colunas].mean(axis=1)

# Exibe apenas a identificação da linha do aluno e as médias calculadas
colunas_resultado = ['ID'] + list(dimensoes.keys())

display(df[colunas_resultado].head().round(2))
```

---

```python
# Média geral das dimensões do FunQ

# Configura o estilo visual do gráfico para o padrão acadêmico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# 1. Calcula a média geral da turma para cada dimensão
medias_dimensoes = df[list(dimensoes.keys())].mean().reset_index()
medias_dimensoes.columns = ['Dimensão', 'Média']

# 2. Cria o gráfico de barras
ax = sns.barplot(
    x='Dimensão', 
    y='Média', 
    data=medias_dimensoes, 
    palette='Blues_d'
)

# 3. Adiciona os valores exatos no topo de cada barra
for p in ax.patches:
    altura = p.get_height()
    ax.annotate(
        f'{altura:.2f}',
        (p.get_x() + p.get_width() / 2., altura),
        ha='center', 
        va='center',
        xytext=(0, 7),
        textcoords='offset points',
        fontsize=11,
        fontweight='bold'
    )

# 4. Ajustes de títulos, eixos e escala Likert (1 a 5)
plt.title('Média Geral das Dimensões do FunQ', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Dimensão do Engajamento', fontsize=12, labelpad=10)
plt.ylabel('Escore Médio (Escala 1 a 5)', fontsize=12)
plt.ylim(1, 5)  # Fixa os limites no padrão da escala Likert do FunQ

# Remove bordas desnecessárias
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
```

---

```python
# Mediana e IQR por dimensão

# Dicionário para armazenar os cálculos de cada dimensão
estatisticas_dimensoes = {}

# Percorre cada dimensão com o laço for
for col in list(dimensoes.keys()):
    # Calcula a mediana
    mediana = df[col].median()
    
    # Calcula o IQR (Q3 - Q1)
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    
    # Salva os resultados no dicionário
    estatisticas_dimensoes[col] = {
        'Mediana': mediana,
        'IQR': iqr
    }

# Transforma o dicionário em um DataFrame do Pandas para exibir como tabela
df_mediana_iqr = pd.DataFrame(estatisticas_dimensoes).T

# Exibe o resultado formatado
display(df_mediana_iqr.round(2))
```

---

```python
# Desvio padrão por dimensão

# Dicionário para armazenar o desvio padrão de cada dimensão
estatisticas_desvio = {}

for col in list(dimensoes.keys()):
    # Calcula o desvio padrão (amostral, ddof=1 é o padrão do pandas)
    desvio_padrao = df[col].std()
    
    estatisticas_desvio[col] = {
        'Desvio Padrão': desvio_padrao
    }

df_desvio_padrao = pd.DataFrame(estatisticas_desvio).T

display(df_desvio_padrao.round(2))
```

---

```python
# Amplitude por dimensão

# Dicionário para armazenar a amplitude (máximo - mínimo) de cada dimensão
estatisticas_amplitude = {}

for col in list(dimensoes.keys()):
    valor_max = df[col].max()
    valor_min = df[col].min()
    amplitude = valor_max - valor_min
    
    estatisticas_amplitude[col] = {
        'Mínimo': valor_min,
        'Máximo': valor_max,
        'Amplitude': amplitude
    }

df_amplitude = pd.DataFrame(estatisticas_amplitude).T

display(df_amplitude.round(2))
```

---

**Cruzamento/correlação entre as dimensões**

Desafio × Imersão → remete à Teoria do Fluxo (Csikszentmihalyi).

Estresse × Prazer → como os itens de Estresse já chegam invertidos (RF04), espera-se correlação positiva (menos estresse, mais prazer).

Autonomia × Prazer → autonomia costuma ser preditor de motivação intrínseca.

```python
# Configura o estilo visual do gráfico para o padrão acadêmico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 7))

# 1. Seleciona apenas as colunas das dimensões
colunas_dim = list(dimensoes.keys())

# 2. Calcula a matriz de correlação entre as dimensões
matriz_correlacao = df[colunas_dim].corr(method='pearson')

# 3. Cria o heatmap
ax = sns.heatmap(
    matriz_correlacao,
    annot=True,            # mostra os valores dentro das células
    fmt='.2f',              # duas casas decimais
    cmap='Blues',            # paleta consistente com o gráfico de médias
    vmin=-1, vmax=1,         # fixa a escala de correlação (-1 a 1)
    square=True,
    linewidths=0.5,
    cbar_kws={'label': 'Coeficiente de Correlação (r)'}
)

# 4. Ajustes de título e eixos
plt.title('Correlação entre as Dimensões do FunQ', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('')
plt.ylabel('')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

---

**Idade como variável exploratória**

```python
# A idade já vem na mesma tabela das respostas (coluna 'Idade' do CSV exportado)
df_idade = df[colunas_dim + ['Idade']].dropna(subset=['Idade']).copy()

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(colunas_dim):
    sns.regplot(
        x='Idade',
        y=col,
        data=df_idade,
        ax=axes[i],
        scatter_kws={'alpha': 0.6, 'color': '#3b6fa0'},
        line_kws={'color': '#08306b'}
    )
    axes[i].set_title(col, fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Idade', fontsize=10)
    axes[i].set_ylabel('Escore (1 a 5)', fontsize=10)
    axes[i].set_ylim(1, 5)

plt.suptitle('Relação entre Idade e as Dimensões do FunQ', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

---

```python
# Médias das Dimensões do FunQ por Faixa Etária

bins = [13, 15, 17]
labels = ['14-15 anos', '16-17 anos']
df_idade['faixa_etaria'] = pd.cut(df_idade['Idade'], bins=bins, labels=labels, right=True)

print(df_idade['faixa_etaria'].value_counts())

medias_por_faixa = df_idade.groupby('faixa_etaria')[colunas_dim].mean().reset_index()

medias_long = medias_por_faixa.melt(
    id_vars='faixa_etaria',
    value_vars=colunas_dim,
    var_name='Dimensão',
    value_name='Média'
)

plt.figure(figsize=(12, 6))
ax = sns.barplot(
    x='Dimensão',
    y='Média',
    hue='faixa_etaria',
    data=medias_long,
    palette='Blues_d'
)

plt.title('Médias das Dimensões do FunQ por Faixa Etária', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Dimensão do Engajamento', fontsize=12, labelpad=10)
plt.ylabel('Escore Médio (Escala 1 a 5)', fontsize=12)
plt.ylim(1, 5)
plt.legend(title='Faixa Etária', bbox_to_anchor=(1.02, 1), loc='upper left')

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

## Análise qualitativa das respostas abertas (q19 a q24)

As questões discursivas são **opcionais**: só as respostas preenchidas entram na análise, e o número de participantes que responderam é registrado por questão.

**Quantidade de participantes que responderam a cada questão discursiva, seguido de seus respectivos percentuais.**

```python
colunas_abertas = {
    'q19_opiniao_atividade': 'q19 - Opinião sobre a atividade',
    'q20_sentimento_jogo': 'q20 - Sentimento durante o jogo',
    'q21_compreensao_conceitos': 'q21 - Compreensão dos conceitos',
    'q22_fator_interesse': 'q22 - Fator de interesse',
    'q23_pontos_negativos': 'q23 - Pontos negativos',
    'q24_sugestoes_mudanca': 'q24 - Sugestões de mudança',
}

def limpar_texto(serie):
    # Remove espaços extras e trata respostas vazias como ausentes
    serie = serie.astype('string').str.strip()
    return serie.mask(serie == '')

abertas = df[list(colunas_abertas)].apply(limpar_texto)

respondentes = pd.DataFrame({
    'Questão': [colunas_abertas[c] for c in abertas.columns],
    'Participantes que responderam': abertas.notna().sum().values,
    '% do total': (abertas.notna().mean() * 100).round(1).values,
})
respondentes
```

---
**Distribuição de quantas questões abertas cada participante respondeu.**

A partir dessas informações, pode-se tirar conclusões a respeito do engajamento com a parte opcional do questionário.

```python
qtd = abertas.notna().sum(axis=1)
distribuicao = (qtd.value_counts().sort_index(ascending=False)
                .rename_axis('Nº de questões abertas respondidas').reset_index(name='Participantes'))
distribuicao['% do total'] = (distribuicao['Participantes'] / len(df) * 100).round(1)
distribuicao
```

**Gráfico dos respondentes por questão**

Mostra quantos alunos responderam a respectiva questão.

```python
ax = respondentes.plot.barh(x='Questão', y='Participantes que responderam', legend=False, figsize=(8, 4))
ax.invert_yaxis()
ax.set_xlabel('Participantes')
ax.set_xlim(0, len(df))
plt.tight_layout()
plt.show()
```

No bloco de código abaixo, substitua os valores pré-definidos pelas categorias desejadas que foram identificadas e definidas após a leitura prévia dos dados qualitativos:

```python
# 2. DEFINA AQUI AS SUAS CATEGORIAS E PALAVRAS-CHAVE
# O dicionário mapeia cada questão a um conjunto de categorias e termos associados.
regras_categorizacao = {
    'q22_fator_interesse': {
        'Material e aspecto físico do jogo': ['peças', 'tabuleiro', 'design', 'material'],
        'Interação em dupla': ['dupla', 'juntos', 'colega'],
        'Perguntas sobre padrões': ['padrões', 'perguntas', 'cartas']
    },
    'q23_pontos_negativos': {
        'Clareza das regras e das perguntas': ['regras', 'confusa', 'pergunta', 'clara'],
        'Tempo insuficiente': ['tempo', 'curto', 'rápido'],
        'Dinâmica entre as duplas': ['duplas', 'atrapalharam', 'concentração']
    },
    'q24_sugestoes_mudanca': {
        'Tempo e organização da atividade': ['tempo', 'aulas', 'organização', 'dividir'],
        'Ampliação e extensão do jogo': ['matérias', 'ampliar', 'outras', 'vezes'],
        'Clareza e dificuldade das regras e perguntas': ['simplificar', 'difícil', 'explicar']
    }
    # Adicione q19, q20, q21 se desejar o mesmo tratamento para elas
}
```

```python
# 3. Função automática para atribuir a categoria com base nas palavras-chave
def atribuir_categoria(texto, regras_da_questao):
    if pd.isna(texto) or str(texto).strip() == '':
        return None
    texto_lower = str(texto).lower()
    for categoria, palavras in regras_da_questao.items():
        if any(palavra in texto_lower for palavra in palavras):
            return categoria
    return 'Outros' # Categoria padrão caso nenhuma palavra-chave bata

# 4. Criação automática das colunas TEMPORÁRIAS em memória (_categoria)
for questao, regras in regras_categorizacao.items():
    col_cat = f"{questao}_categoria"
    if questao in df.columns:
        df[col_cat] = df[questao].apply(lambda x: atribuir_categoria(x, regras))

# 5. Mapeamento dos rótulos para exibição no relatório
colunas_abertas = {
    'q19_opiniao_atividade': 'Questão 19',
    'q20_sentimento_jogo': 'Questão 20',
    'q21_compreensao_conceitos': 'Questão 21',
    'q22_fator_interesse': 'Questão 22',
    'q23_pontos_negativos': 'Questão 23',
    'q24_sugestoes_mudanca': 'Questão 24'
}

# 6. colunas temporárias da memória
for coluna, rotulo in colunas_abertas.items():
    col_cat = coluna + '_categoria'      # ex.: q22_categoria
    if col_cat not in df.columns:
        continue                         # questão sem regras ou não presente no dataframe

    sub = pd.DataFrame({
        'resposta': df[coluna],
        'categoria': df[col_cat].astype('string').str.strip().mask(lambda s: s == ''),
    }).dropna(subset=['resposta'])
    
    sub['categoria'] = sub['categoria'].str.capitalize()   # ignora diferenças de maiúsculas

    tabela = (
        sub['categoria'].value_counts()
        .rename_axis('Categoria')
        .reset_index(name='Frequência')
    )
    tabela['% das respostas'] = (tabela['Frequência'] / len(sub) * 100).round(1)
    
    if not sub.empty and 'categoria' in sub.columns:
        exemplos = sub.groupby('categoria')['resposta'].agg(lambda s: s.value_counts().index[0])
        tabela['Exemplo de resposta'] = tabela['Categoria'].map(exemplos)
    else:
        tabela['Exemplo de resposta'] = ''

    sem_categoria = sub['categoria'].isna().sum()
    print(f"{rotulo}  (n = {len(sub)} respostas; sem categoria: {sem_categoria})")
    display(tabela)
    print()
```
